Break : Stop a Loop early.

```python
for i in range(1,10):
  if i == 5:
    break
  print(i)
```

```python
for i in range(1,10):
  if i == 5:
    continue
  print(i)
```

Continue: Skip one turn


```python
for i in range(1,6):
  if i == 3:
    continue
  print(i)
```
