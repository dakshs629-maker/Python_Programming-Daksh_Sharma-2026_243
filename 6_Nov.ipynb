Swap two values using third variable.

```python
p,q = (1,2)
c= p+q
print("c:",c)
p= c-p
print("p:",p)
q= c-p
print("q:",q)
print("(p,q)=",p,q)
```

```python
p,q = 1,2
q=p
print(p,q)
p=q
print(p,q)
```
