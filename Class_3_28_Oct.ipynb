```python
a= 50
b= 60.5
c= "Hola"
d= True
e= [1,4,8]
f= (3,5,8)
g= {1,5,7}
h= {"name":"Mona", "Age": 20}

import sys
print("Python version is",sys.version)
print("Type of a is", type(a).__name__)
print("Type of b is", type(b))
print("Type of c is", type(c))
print("Type of d is", type(d))
print("Type of e is", type(e))
print("Type of f is", type(f))
print("Type of g is", type(g))
print("Type of h is", type(h))
print("")

print("Size of Integer is",sys.getsizeof(a), "bytes.")
print("Size of Float is",sys.getsizeof(b), "bytes.")
print("Size of String is",sys.getsizeof(c), "bytes.")
print("Size of Boolean is",sys.getsizeof(d), "bytes.")
print("Size of List is",sys.getsizeof(e), "bytes.")
print("Size of Tuple is",sys.getsizeof(f), "bytes.")
print("Size of Set is",sys.getsizeof(g), "bytes.")
print("Size of Dictionary is",sys.getsizeof(h), "bytes.")
```
